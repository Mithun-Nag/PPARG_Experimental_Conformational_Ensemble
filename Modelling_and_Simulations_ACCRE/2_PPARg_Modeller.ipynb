## 🧬 Batch Reconstruction of Missing Residues with MODELLER

🔹 Each curated **PPARγ X-ray chain** is used independently as a structural template.  
🔹 The **canonical PPARγ LBD sequence** is aligned to each template to reconstruct residues missing from the experimental structure.  
🔹 **One full-atom model** is generated for each template while preserving the available experimental coordinates.  
🔹 Residues are **renumbered consistently from 1** across all modeled structures.  
🔹 Final models retain the original **PDB ID and chain ID** filenames and are saved in `MODELLER_MODELS/`.  
🔹 A complete **MODELLER run log** is recorded for reproducibility and troubleshooting.

```python
# ============================================================
# PPARγ MODELLER BATCH HOMOLOGY MODELING
# ============================================================
# - One model per X-ray template
# - Full-atom templates
# - Canonical PPARG LBD sequence (already defined)
# - Uniform residue numbering
# - Output visible in Jupyter AND written to log
# ============================================================

# -------------------------
# Imports
# -------------------------
import sys
from pathlib import Path
from modeller import *
from modeller.automodel import *

# ============================================================
# TEE LOGGER: print to Jupyter + file simultaneously
# ============================================================
class Tee:
    def __init__(self, *streams):
        self.streams = streams

    def write(self, data):
        for s in self.streams:
            s.write(data)
            s.flush()

    def flush(self):
        for s in self.streams:
            s.flush()

# ============================================================
# OUTPUT DIRECTORY + LOG FILE
# ============================================================
outdir = Path("MODELLER_MODELS")
outdir.mkdir(exist_ok=True)

logfile = outdir / "run.log"
log_fh = open(logfile, "w")

# Redirect stdout/stderr through Tee
sys.stdout = Tee(sys.__stdout__, log_fh)
sys.stderr = sys.stdout

print("=== MODELLER batch run started ===")

# ============================================================
# INPUT: template directory
# ============================================================
template_dir = Path("PPARG_CHAINS_PDB_all")
templates = sorted(template_dir.glob("*.pdb"))

assert templates, "❌ No PDB templates found"

print(f"Found {len(templates)} PPARG templates")

# ============================================================
# MODELLER environment
# ============================================================
env = Environ()
env.io.atom_files_directory = ['.', str(template_dir)]

# ============================================================
# MODELLER class
# ============================================================
class PPARGModel(AutoModel):
    """
    Custom MODELLER class:
    - enforce single chain
    - renumber residues from 1
    """
    def special_patches(self, aln):
        self.rename_segments(
            segment_ids=[self.chains[0].name],
            renumber_residues=1
        )

# ============================================================
# MAIN LOOP: one template → one model
# ============================================================
for pdb in templates:
    name = pdb.stem
    print(f"\n▶ Modeling from template: {name}")

    # --------------------------------------------------------
    # Build alignment (canonical MODELLER workflow)
    # --------------------------------------------------------
    aln = Alignment(env)

    # Read template sequence directly from PDB
    mdl = Model(env, file=name)
    aln.append_model(mdl, align_codes=name)

    # Append target PPARG sequence
    aln.append_sequence(PPARG_SEQ)
    aln[-1].code = "PPARG"

    alnfile = f"{name}.ali"
    aln.write(file=alnfile, alignment_format="PIR")

    # --------------------------------------------------------
    # Build homology model
    # --------------------------------------------------------
    a = PPARGModel(
        env,
        alnfile=alnfile,
        knowns=name,
        sequence="PPARG"
    )

    a.starting_model = 1
    a.ending_model   = 1
    a.make()

    # --------------------------------------------------------
    # Collect output
    # --------------------------------------------------------
    raw_model = Path("PPARG.B99990001.pdb")
    final_model = outdir / pdb.name  # preserve template name

    if raw_model.exists():
        raw_model.rename(final_model)
        print(f"  ✔ Saved {final_model.name}")
    else:
        print(f"  ✖ Model not generated for {name}")

    # Move MODELLER per-model logs
    for log in Path(".").glob("PPARG*.log"):
        log.rename(outdir / f"{name}_{log.name}")

print("\n=== MODELLER batch run complete ===")
print(f"Full run log: {logfile.resolve()}")

# ============================================================
# CLEANUP
# ============================================================
log_fh.close()
sys.stdout = sys.__stdout__
sys.stderr = sys.__stderr__
```
